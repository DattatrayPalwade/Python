```python
Problem statement : Bank Transaction Summary

A bank stores customer transactions as a list of dictionaries. 
Each transaction contains an account number, transaction type, and amount.

Write a Python program to calculate the total deposits, total withdrawals, 
and net amount for each account.

Net amount = Total deposits − Total withdrawals

Sample input --->
transactions = [
    {"account": "A101", "type": "deposit", "amount": 5000},
    {"account": "A102", "type": "deposit", "amount": 8000},
    {"account": "A101", "type": "withdrawal", "amount": 1200},
    {"account": "A101", "type": "deposit", "amount": 2000},
    {"account": "A102", "type": "withdrawal", "amount": 3000},
    {"account": "A103", "type": "withdrawal", "amount": 500},
    {"account": "A102", "type": "deposit", "amount": 1500}
]

Expected output --->
Account: A101
Deposits: 7000
Withdrawals: 1200
Net Amount: 5800

Account: A102
Deposits: 9500
Withdrawals: 3000
Net Amount: 6500

Account: A103
Deposits: 0
Withdrawals: 500
Net Amount: -500
```

```python
# Transactions
transactions = [
    {"account": "A101", "type": "deposit", "amount": 5000},
    {"account": "A102", "type": "deposit", "amount": 8000},
    {"account": "A101", "type": "withdrawal", "amount": 1200},
    {"account": "A101", "type": "deposit", "amount": 2000},
    {"account": "A102", "type": "withdrawal", "amount": 3000},
    {"account": "A103", "type": "withdrawal", "amount": 500},
    {"account": "A102", "type": "deposit", "amount": 1500}
]

# Summary
summary = {}

for t in transactions :
    account = t["account"]
    account_type = t["type"]
    amount = t["amount"]

    # Initialize
    if account not in summary :
        summary[account] = { "deposits" : 0 , "withdrawals" : 0 }

    if account_type == "deposit" :
        summary[account]["deposits"] += amount
    else :
        summary[account]["withdrawals"] += amount

# final summary
for account , account_transactions in summary.items():  # Only keys : .keys() and Only values : .values()
    net_amount =  account_transactions["deposits"] - account_transactions["withdrawals"]

    print(f"Account : {account}")
    print(f"Deposits : {account_transactions["deposits"]}")
    print(f"Withdrawals : {account_transactions["withdrawals"]}")
    print(f"Net Amount : {net_amount}")
    print("----------------------------------------------------------------------")
    
```
