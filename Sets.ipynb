```python
"""============================================================================
Creating sets:
============================================================================"""
# 1. multiple elements
numbers = { 10 , 20 , 30 , 40 , 50 }
print(numbers)

# 2. multiple data type elements
student = { "Dattatray" , 21 , 600 , "Yes" }
print(student)

# 3. Empty tuple
e = set()
print(type(e))
```

```python
"""============================================================================
Taking sets input:
============================================================================"""
# 1. Take each value seperately
name = input("Enter name : ") 
age = int(input("Enter age : "))

student = { name , age }
print(student)
print(type(student))

# 2. Take multiple values
subject = set( map( str , input("Enter subjects : ").split(",")))
print(subject)

# 3. Taking multiple inputs using loop
product_list = set()
n = int(input("Enter number of products : "))

for i in range(n) :
    product = input(f"Enter { i + 1 } product name : ")
    product_list.add(product)

print(product_list)
    
```

```python
"""============================================================================
Set methods:
- add()      : Add elements                               : set.add(value)
- update()   : Adds elements from one or more iterables   : set.update(iterable)
- remove()   : Removes an element                         : set.remove(value)
- discard()  : Removes an element if it exists            : set.discard(value)
- pop()      : Remove and return an element               : set.pop()
- clear()    : Remove all elements                        : set.clear()
- copy()     : creates a copy                             : new_set = set.copy()
============================================================================"""

s = { 10 , 20 , 30 , 40 , 50 }
print(s)

s.add(60)
print(s)

s.update([100 , 200])
print(s)

s.remove(30)
print(s)

s.discard(210)
print(s)

s.clear()
print(s)

s = { 10 , 20 , 30 , 40 , 50 }
new_s = s.copy()
print(new_s)

s = { 10 , 20 , 30 , 40 , 50 }
new_s = s.pop()
print(new_s)
print(s)
```
