```python
"""
*******************************************************************************
Filename  : Format.py
Author    : Dattatray Palwade - Value Tech Academy
*******************************************************************************

===============================================================================
Dictionary:
===============================================================================
A dictionary stores data in {key : value} pairs. Keys are unique values can be of
different data types.
Dictionaries are mutable.

Syntax --->
    dict_name = { key : value , key : value , ... }

"""
```

```python
"""============================================================================
Create dict:
============================================================================"""
# 1. Basic dictionary
student = {
    "roll_no" : 28 ,
    "name" : "Dattatray" ,
    "age" : 21 
}
print(student)

# 2. Empty dict
student = {}
print(type(student))

# 3. Using for loop
n = int(input("Enter number of entries : "))

for i in range(n) :
    key = input(f"Enter key { i + 1 } : ")
    value = input(f"Enter value for { i + 1 } key : ")
    student[key] = value

print(student)
```

```python
"""============================================================================
Output dict:
============================================================================"""
student = {
    "roll_no" : 28 ,
    "name" : "Dattatray" ,
    "age" : 21 
}

# Single output
roll_no = student["roll_no"]
print(roll_no)

print("-----------------------------------------")
# Using for loop
#  1. By using key
student = {
    "roll_no" : 28 ,
    "name" : "Dattatray" ,
    "age" : 21 
}

for key in student:
    print(f" {key}  : {student[key]} " )

print("-----------------------------------------")
# 2. By using items()
for key , value in student.items() :
    print(f"{key}   : {value} ")


print("-----------------------------------------")
# 3. By using keys()
for key in student.keys():
    print(key)

print("-----------------------------------------")
# 4. By using values()
for value in student.values() :
    print(value)
```
