```python
"""============================================================================
Bubble sort:
Working :

Example --->

Array - 10 , 60 , 30 , 40 , 50 , 20

step1 - 10 , 60 , 30 , 40 , 50 , 20 ( No swap )
step2 - 10 , 30 , 60 , 40 , 50 , 20 ( index 1 swap with index 2 )
step3 - 10 , 30 , 40 , 60 , 50 , 20 ( index 2 swap with index 3 )
step4 - 10 , 30 , 40 , 50 , 60 , 20 ( index 3 swap with index 4 )
step5 - 10 , 30 , 40 , 50 , 20 , 60 ( index 4 swap with index 5 )
step6 - 10 , 30 , 40 , 20 , 50 , 60
step7 - 10 , 30 , 20 , 40 , 50 , 60
step8 - 10 , 20 , 30 , 40 , 50 , 60 ( completely sorted )
 

Repeat the steps starting from the start until no items are swapped.
============================================================================"""

###############################################################################
####################### ADD YOUR IMPLEMENTATION HERE ##########################
###############################################################################

# Taking list as a input
li = list(map( int , input("Enter numbers : ").split()))
print(f"Starting array : {li}")

length = len(li) # Lenght of array

# Outer loop - for number of iterations
for i in range(length) :
    # Inner loop - for comparison
    for j in range( 0 , length - 1 - i ) :
        # compare adjacent elements
        if li[j] > li[ j + 1 ] :
            # swap
            li[j] , li[ j + 1 ] = li[ j + 1 ] , li[j]

print(f"Sorted array : {li} ")

###############################################################################
######################## END OF YOUR IMPLEMENTATION ###########################
###############################################################################


######################### DO NOT EDIT BELOW THIS LINE #########################
```

```python
"""============================================================================
Selection sort:
Working :

Example --->

Array - 20 , 12 , 10 , 15 , 2

step1 - ( index 0 ) 20 , 12 , 10 , 15 , 2  : min value at index 1
step2 - ( index 1 ) 20 , 12 , 10 , 15 , 2  : min value at index 2
step3 - ( index 2 ) 20 , 12 , 10 , 15 , 2  : min value at index 2
step4 - ( index 3 ) 20 , 12 , 10 , 15 , 2  : min value at index 4
step5 - after swap : 2 , 12 , 10 , 15 , 20
============================================================================"""


###############################################################################
####################### ADD YOUR IMPLEMENTATION HERE ##########################
###############################################################################

# Taking list as a input
li = list(map( int , input("Enter numbers : ").split()))
print(f"Starting array : {li}")

length = len(li)

# Outer loop 
for i in range(length) :
    min_index = i
    
    for j in range( i + 1 , length ) :

        if li[j] < li[min_index] :
            min_index = j

    li[i] , li[min_index] = li[min_index] , li[i]

print(f"Sorted array : {li}")

###############################################################################
######################## END OF YOUR IMPLEMENTATION ###########################
###############################################################################


######################### DO NOT EDIT BELOW THIS LINE #########################
```
