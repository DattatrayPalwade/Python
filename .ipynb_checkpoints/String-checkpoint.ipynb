```python
'''
Python strings are immutable --->
Once a string object is created, its charcters cannot be changed in place.
'''

# Example : 
s = "Good"
s[0] = "F"

# We will get error because strings are immutable
# TypeError: 'str' object does not support item assignment
```

```python
# Modify strings --->

# Example 

s = "Good"
s = "F" + s[1:]
print(s)
```

```python
'''
Slicing ---> It is one of the most important python string features
Syntax ---> 
    string[start : end]

start ---> Included ( start with index 0 )
end ---> Excluded ( end with index n + 1 )
'''

# Example 1 : slicing of string
s = "Dattatray"
s = s[0:9]
print(s)

print("-------------------------------------------------------------------------")
# Example 2 : slicing from beginning
s = "Dattatray"
print(s[:5])

print("-------------------------------------------------------------------------")
# Example 2 : slicing from index to the end
s = "Dattatray"
print(s[5:])
```
