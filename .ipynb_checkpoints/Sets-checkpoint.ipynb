```python
"""============================================================================
Creating sets:
============================================================================"""
# 1. multiple elements
numbers = { 10 , 20 , 30 , 40 , 50 }
print(numbers)

# 2. multiple data type elements
student = { "Dattatray" , 21 , 600 , "Yes" }
print(student)

# 3. Empty tuple
e = set()
print(type(e))
```

```python
"""============================================================================
Taking sets input:
============================================================================"""
# 1. Take each value seperately
name = input("Enter name : ") 
age = int(input("Enter age : "))

student = { name , age }
print(student)
print(type(student))

# 2. Take multiple values
subject = set( map( str , input("Enter subjects : ").split(",")))
print(subject)
```

```python
"""============================================================================
Accessing tuple elements:
============================================================================"""
# 1. Indexing
n = ( 10 , 20 , 30 , 40 , 50 )
print(n[0])

for index in range(len(n)) :
    print(n[index])

# 2. Slicing
n = ( 10 , 20 , 30 , 40 , 50 )
print(n[0: 3])

# 3. Unpack and Display
employee = ( "Dattatray" , 10000 , "IT" )

Employee_Name , Employee_Salary , Employee_Dept = employee
print(f"Employee Name : {Employee_Name} ")
print(f"Employee Salary : {Employee_Salary} ")
print(f"Employee Dept : {Employee_Dept} ")
```

```python
"""============================================================================
tuple methods and functions:
- count()    : counts occurences of a value
- index()    : Return first match value index
- len()      : Returns a lenght of tuple
- max()      : Returns maximum value
- min()      : Returns minimum value
- sorted()   : sort tuple values
============================================================================"""
# 1. count() method ---> tuple_name.count(value) 
n = ( 10 , 20 , 30 , 20 , 50 )
print(n.count(20))

# 2. index() method ---> tuple_name.index(value)
n = ( 10 , 20 , 30 , 20 , 50 )
print(n.index(20))

# 3. len() Function ---> len(tuple)
n = ( 10 , 20 , 30 , 20 , 50 )
print(len(n))

# 4. max() Function ---> max(tuple)
n = ( 10 , 20 , 30 , 20 , 50 )
print(max(n))

# 5. min() Function ---> min(tuple)
n = ( 10 , 20 , 30 , 20 , 50 )
print(min(n))

# 6. sorted() Function ---> sorted(tuple)
n = ( 10 , 20 , 30 , 20 , 50 )
print(sorted(n)) # sorted data output will be in list data type
print(tuple(sorted(n)))  # Typecast to convert into tuple
```
