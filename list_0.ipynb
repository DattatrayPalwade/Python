```python
"""============================================================================
Creating lists: 
Syntax --->
    list_name = [item1 , item2 , item3 , item4]
============================================================================"""
# 1. List with values
electronic_devices = [ "Laptop", "T.V", "Desktop", "Mixer", "Printer" ]
print(electronic_devices)

# 2. Empty list
li = []
print(li)

# 3. Using list()
name = "Dattatray"
name_list = list(name)
print(name_list)

text_msg = "Shree Swami Samarth !..."
msg_list = list(map( str , text_msg.split(" ")))
print(msg_list)

# 4. Repeated values
zeros = [0] * 2
print(zeros)
```

```python
"""============================================================================
Display lists:
============================================================================"""
# 1. Using print() Function
li = [ 10 , 20 , 30 , 40 , 50 ]
print(li)

# 2. Iterating with a for Loop
li = [ 10 , 20 , 30 , 40 , 50 ]

for index in li :
    print(index)

# 3. Using the Unpacking Operator (*)
li = [ 10 , 20 , 30 , 40 , 50 ]
print(*li)
print( *li , sep="\n" ) 

# 4. Using the .join() String Method
li = [ 10 , 20 , 30 , 40 , 50 ]
print(",".join(map( str , li)))

# 5. Using enumerate() for Indexed Display
electronic_devices = [ "Laptop", "T.V", "Desktop", "Mixer", "Printer" ]

for index, value in enumerate(electronic_devices):
    print(f"{index} = {value}")
```

```python
"""============================================================================
Taking List as Input:
============================================================================"""
# 1. Using input() Function
li = list(map( int , input("Enter numbers : ").split(",")))
print(li)

# 2. Single line string
s_list = input("Enter product name : ").split(",")
print(s_list)

# 3. Using for loop
n = int(input("Enter number of elements : "))
n_list = []

for i in range(n) :
    number = int(input(f"Enter number { i + 1 } : "))
    n_list.append(number)

print(n_list)
```

```python
"""============================================================================
Indexing: It is used to access elements from the list.
Syntax ---> 
    list_name[index]
============================================================================"""
# 1. Forward indexing ( Starting index ---> 0 , End index ---> n-1 )
li = [ 10 , 20 , 30 , 40 , 50 ]
index = li[0]
print(index)

# 2. Reverse indexing ( Starting index ---> -1 , End index ---> -n )
li = [ 10 , 20 , 30 , 40 , 50 ]
index = li[-2]
print(index)

# 3. Updating list elements
li = [ 10 , 20 , 30 , 40 , 50 ]
li[2] = 300
print(li)
```

```python
"""============================================================================
List Slicing:
Syntax ---> 
    list_name[ start : end : step ]

- start is inclusive ( means it include the elements from start index )
- end is exclusive ( means it exclude the elements from end index )
============================================================================"""
# Example:
li = [ 10 , 20 , 30 , 40 , 50 , 60 ]
li_slice = li[ 1 : 4 ]
print(li_slice)

# 1. list_name[ start : ]
li = [ 10 , 20 , 30 , 40 , 50 , 60 ]
li_slice = li[ 1 : ]
print(li_slice)

# 2. list_name[  : end ]
li = [ 10 , 20 , 30 , 40 , 50 , 60 ]
li_slice = li[ : 4 ]
print(li_slice)

# 3. list_name[ : ]
li = [ 10 , 20 , 30 , 40 , 50 , 60 ]
li_slice = li[ : ]
print(li_slice)

# 4. list_name[ : : step ]
li = [ 10 , 20 , 30 , 40 , 50 , 60 ]
li_slice = li[ : : 2 ]
print(li_slice)

# 5. reverse order
li = [ 10 , 20 , 30 , 40 , 50 , 60 ]
li_slice = li[ : : -1 ]
print(li_slice)
```

```python
"""============================================================================
Adding elements in list:
- append()    : list_name.append(value)
- insert()    : list_name.insert( index , value )
- extend()    : list_name.extend(iterable)
============================================================================"""
# 1. append() Function : adds elements to the end
li = [ 10 , 20 , 30 , 40 , 50 ]
li.append(60)
print(li)

li = [ 10 , 20 , 30 , 40 , 50 ]
lst = [ 60 , 70 , 80 , 90 , 100 ]
li.append(lst)
print(li)

# 2. insert() Function : adds element at a particular position
li = [ 10 , 20 , 30 , 40 , 50 ]
li.insert(3 , 60)
print(li)

# 3. extend() Function : adds multiple elements from another iterable ( list )
li = [ 10 , 20 , 30 , 40 , 50 ]
lst = [ 60 , 70 , 80 , 90 , 100 ]
li.extend(lst)
print(li)
```

```python
"""============================================================================
Removing elements in list:
- remove()    : list_name.remove(value)
- pop()       : list_name.pop() / list_name.pop(index)
- del         : del list_name[index / slice]
- clear()     : list_name.clear()
============================================================================"""
# 1. remove() Function : Removes the first matching value
numbers = [ 10 , 20 , 30 , 40 , 50 ] 
numbers.remove(20)
print(numbers)

# 2. pop() Function : Removes and returns an element
numbers = [ 10 , 20 , 30 , 40 , 50 ] 
numbers.pop(4)
print(numbers)

# 3. del : deletes using an index or slice
numbers = [ 10 , 20 , 30 , 40 , 50 ] 
del numbers[1]
print(numbers)

# 4. clear() Function : Removes everything
numbers = [ 10 , 20 , 30 , 40 , 50 ] 
numbers.clear()
print(numbers)
```

```python
"""============================================================================
Finding and Counting Elements:
- index()    : list_name.index(value)
- count()    : list_name.count(value)
============================================================================"""
# 1. index() Function : It retruns first matching index
names = [ "Laptop" , "Mouse" , "Keyboard" , "Printer" ]
index = names.index("Mouse")
print(index)

# 2. count() Function : Counts the repetation of an value
numbers = [ 10 , 20 , 10 , 50 , 60 , 10 , 100 , 500 ]
count = numbers.count(10)
print(count)
```

```python
"""============================================================================
Membership Operators:
- in
- not in

Operator ---> It is a special symbol which use to perform operation on operands 
Operands ---> The values on which the operation is perform 
============================================================================"""
# 1. in operator
names = [ "Laptop" , "Mouse" , "Keyboard" , "Printer" ]
print( "Mouse" in names )

# 2. not in operator
names = [ "Laptop" , "Mouse" , "Keyboard" , "Printer" ]
print( "CPU" not in names )
```

```python
"""============================================================================
Built-in Functions with lists: 
- len()
- sum()
- min()
- max()
- sorted()
============================================================================"""
numbers = [ 100 , 600 , 50 , 60 , 12 , 64 ]

print(f" Lenth of list : {len(numbers)}")
print(f" Sum of all elements in list {sum(numbers)}")
print(f" Minimum number from list {min(numbers)}")
print(f" Maximum number from lsit {max(numbers)}")
print(f" Sorted list {sorted(numbers)}")
```

```python
"""============================================================================
Reversing a list:
- reverse()
- sort()
============================================================================"""
# 1. sort() Function
numbers = [ 100 , 600 , 50 , 60 , 12 , 64 ]
numbers.sort(reverse = True)
print(numbers)

# 2. reverse() Function
numbers = [ 100 , 600 , 50 , 60 , 12 , 64 ]
numbers.reverse()
print(numbers)
```

```python
"""============================================================================
Copying list:
============================================================================"""
# 1. By using assignment operator
li = [ 10 , 20 , 30 , 40 , 50 ]
lst = li

print(li)
print(lst)

# 2. copy() Function
li = [ 10 , 20 , 30 , 40 , 50 ] 
lst = li.copy()

li.append(500)
print(li)
print(lst)
```
